```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

### 1. Change the unit of Revenue to \\$ million. Divide the data into 75% training and 25% test set (use random_state=15) and run the following linear regression models on training data. Report coefficients (UK, US are country dummy variables).
- M1: Revenue = β0 + β1 Budget + β2 Rating + β3 UK + β4 US + e
- M2: Revenue = β0 + β1 rating + β2 UK + β3 US + e

```python
# Read data

df = pd.read_csv('Movie.csv')
df.head()
```

```python
# Change the unit of Revenue to $ million

df['revenue'] = df.revenue/1000000  # Change Revenue to $ million unit
```

- `df.revenue`/1000000`: Divides each value in the 'rev' column by one million.
- `df['revenue']`: The result of the division is assigned back to the 'rev' column in the DataFrame.

```python
# Creating dummy variables for 'country' 

cntry_dum = pd.get_dummies(df.country, dtype=int)
  #  `dtype=int` ensures that the dummy variables are in numeric form (1 and 0).

# Adding the dummy variables to the original dataframe

df = pd.concat([df, cntry_dum], axis=1)
df.head()
```

```python
# Define x, y. Divied the date into training and test sets. 

x = df[['budget', 'rating', 'UK', 'US']]
y = df.revenue
xtrain1, xtest1, ytrain, ytest = train_test_split(x, y, random_state = 15)  # default: 75%, 25% split
```

- `x = df[[  ]]`: Creates a DataFrame x by selecting specific columns from the original DataFrame df.
- `y = df.revenue`: Sets the target variable y to be the 'revenue' column from the original DataFrame df.
- `xtrain1, xtest1, ytrain, ytest = train_test_split(x, y, random_state = 15)`: This line uses the train_test_split function to split your data into training and testing sets for regression.
    - `random_state = 15`: Setting the random_state parameter to 15 ensures that the data is split the same way each time you run the code, providing reproducibility in your results.
- `len(xtrain1)`: Calculates and returns the number of samples (data points) in your training set (xtrain1). It gives you the count of rows in the training data.

```python
# Define x for model 2

xtrain2 = xtrain1.drop('budget', axis=1)
xtest2 = xtest1.drop('budget', axis=1)
```

- `xtrain2 = xtrain1.drop( )`: Creates a new DataFrame xtrain2 by dropping the column from the xtrain1 DataFrame. The axis=1 argument indicates that you're removing a column (i.e., a feature).
- `xtest2 = xtest1.drop( )`: Similarly, this line creates a new DataFrame xtest2 by dropping the column from the xtest1 DataFrame.

```python
# Estimate model 1, 2

m1 = LinearRegression().fit(xtrain1, ytrain)
m2 = LinearRegression().fit(xtrain2, ytrain)
```

- `m1 = LinearRegression().fit(xtrain1, ytrain)`:
    - `LinearRegression()`: Initializes a linear regression model.
    - `fit(xtrain1, ytrain)`: Fits the linear regression model to the training data.

```python
# Obtain coefficients and intercept

print('m1 coefficents', m1.coef_)
print('m2 coefficents', m2.coef_)
print('m1 intercept', m1.intercept_)
print('m2 intercept', m2.intercept_)
```

- `m1.coef_` retrieves the coefficients of the linear regression model m1. The coefficients represent the weights assigned to each feature (independent variable) in the model and are used to make predictions.
- `m1.intercept_` retrieves the intercept term of the linear regression model m1. The intercept is also referred to as the bias or constant term, and it represents the predicted value of the target variable when all the feature values are zero.

### 2. Calculate MSE, RMSE, MAE, and R-squared from the two models.

```python
# Predict y on test data

pred1 = m1.predict(xtest1)
pred2 = m2.predict(xtest2)
```

- `pred1 = m1.predict(xtest1)`: Uses the trained linear regression model m1 to make predictions on the test data xtest1, and the predictions are stored in the variable pred1.

```python
# Obtain MSE, RMSE, MAE, R2 for M1

print('mse1', mean_squared_error(ytest, pred1))
print('rmse1', mean_squared_error(ytest, pred1)**0.5)
print('mae1', mean_absolute_error(ytest, pred1))
print('R-square1', r2_score(ytest, pred1))
```

- `mean_squared_error(ytest, pred1)`: Calculates the Mean Squared Error (MSE) for model m1. The MSE measures the average of the squared differences between the true target values (ytest) and the predicted values (pred1).
- `mean_squared_error(ytest, pred1)**0.5`: Calculates the Root Mean Squared Error (RMSE) for model m1. The RMSE is the square root of the MSE and provides a measure of the average error in the same units as the target variable.
- `mean_absolute_error(ytest, pred1)`: Calculates and prints the Mean Absolute Error (MAE) for model m1. The MAE measures the average absolute differences between the true target values and the predicted values.
- `r2_score(ytest, pred1)`: Calculates and prints the R-squared (R^2) score for model m1. The R^2 score measures the proportion of the variance in the target variable (ytest) that is explained by the model's predictions (pred1).

```python
# Obtain MSE, RMSE, MAE, R2 for M2

print('mse2', mean_squared_error(ytest, pred2))
print('rmse2', mean_squared_error(ytest, pred2)**0.5)
print('mae2', mean_absolute_error(ytest, pred2))
print('R-square2', r2_score(ytest, pred2))
```

#### Model 1 shows better predictive performance than Model 2
- Lower error measures (MSE, RMSE, MAE) and higher R-squared

```python
#  Alternative

error1 = pred1 - ytest
mse1 = (error1**2).mean()
rmse1 = (error1**2).mean()**0.5
mae1 = abs(error1).mean()

print('mse1:', mse1)
print('rmse1:', rmse1)
print('mae1:', mae1)
```

- `error1 = pred1 - ytest` calculates the errors by subtracting the true target values (ytest) from the predicted values (pred1).
- `mse1 = (error1**2).mean()` calculates the Mean Squared Error (MSE) by squaring each error, averaging those squared errors, and taking the mean.
- `rmse1 = (error1**2).mean()**0.5` calculates the Root Mean Squared Error (RMSE). It's the square root of the MSE.
- `mae1 = abs(error1).mean()` calculates the Mean Absolute Error (MAE) by taking the absolute values of the errors, averaging them, and calculating the mean.

```python
error2 = pred2 - ytest
mse2 = (error2**2).mean()
rmse2 = (error2**2).mean()**0.5
mae2 = abs(error2).mean()

print('mse2:', mse2)
print('rmse2:', rmse2)
print('mae2:', mae2)
```

### 3. Run 5-fold cross validations and calculate MSEs, MAEs, and R-squares for the two models.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn import model_selection
from sklearn.model_selection import KFold
```

```python
#  Define linear regression and a 5-fold CV

m = LinearRegression()
cv5 = model_selection.KFold(n_splits=5) 
```

- `m = LinearRegression()`: This line initializes a linear regression model.
- `cv5 = model_selection.KFold(n_splits=5)`: This line creates a 5-fold cross-validation scheme. 
    - This can be omitted for 5-fold CV (default)

```python
# Define x, y for M1

y = df.revenue
x1 = df[['budget', 'rating', 'UK', 'US']]
```

- `y = df.rev`: y is the dependent variable or target variable. In this case, it is set to the 'rev' column of the DataFrame df.
- `x1 = df[['sequel', 'avgscreen', 'budget', 'rating', 'ad']]`: x1 is a DataFrame that includes a set of independent variables or features that you want to use in your regression analysis.

```python
# Obtain MSE, MAE, R2 for M1

mse1 = -(model_selection.cross_val_score(m, x1, y, cv=cv5, scoring='neg_mean_squared_error'))
mae1 = -(model_selection.cross_val_score(m, x1, y, cv=cv5, scoring='neg_mean_absolute_error'))
rsq1 = model_selection.cross_val_score(m, x1, y, cv=cv5, scoring='r2')
```

- `mse1 = -(model_selection.cross_val_score(m, x1, y, cv=cv5, scoring='neg_mean_squared_error'))`: Calculates the Mean Squared Error (MSE) using 5-fold cross-validation for the linear regression model m.
    - `model_selection.cross_val_score` performs cross-validation
    - `scoring='neg_mean_squared_error'` indicates that it should use MSE as the evaluation metric. The negative sign is used to obtain a positive MSE value.
    - cv=cv5 can be omitted for 5-fold (default)
- `mae1 = -(model_selection.cross_val_score(m, x1, y, cv=cv5, scoring='neg_mean_absolute_error'))`: This line calculates the Mean Absolute Error (MAE) using 5-fold cross-validation for the linear regression model m.
- `rsq1 = model_selection.cross_val_score(m, x1, y, cv=cv5, scoring='r2')`: This line calculates R-squared using 5-fold cross-validation for the linear regression model m. 

```python
# Scores of each run of CV

print('mse1', mse1)
print('mae1', mae1)
print('rsq1', rsq1)
```

```python
# Average scores of all the runs of CV

print('mse1', mse1.mean())
print('mae1', mae1.mean())
print('r-square1', rsq1.mean())
```

- `mean()`: This code computes the mean value of the MSE, MAE, and R-squared scores across the different folds of cross-validation.

```python
# Define x for M2

x2 = x1.drop('budget', axis=1)
# x2 = df[['rating', 'UK', 'US']]  # Alternative
```

 - `x2 = x1.drop('ad', axis=1)` is creating a new DataFrame x2 by removing the 'ad' column from the existing DataFrame x1
 - `axis=1` indicates that you are removing a column.

```python
# Obtain MSE, MAE, R2 for M2

mse2=-(model_selection.cross_val_score(m, x2, y, cv=cv5, scoring='neg_mean_squared_error'))
mae2=-(model_selection.cross_val_score(m, x2, y, cv=cv5, scoring='neg_mean_absolute_error'))
rsq2=model_selection.cross_val_score(m, x2, y, cv=cv5, scoring='r2')
```

```python
# scores of each run of CV

print(mse2)
print(mae2)
print(rsq2)
```

```python
# Average scores of all the runs of CV

print('mse2', mse2.mean())
print('mae2', mae2.mean())
print('r-square2', rsq2.mean())
```

- Model 1 shows better predictive performance than Model 2: Model 1 has lower error measures (MSE, MAE) and higher R-squared
